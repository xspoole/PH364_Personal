```python
import numpy as np
from numpy.polynomial import Legendre, Polynomial
import math
import sympy as sp
import matplotlib.pyplot as plt
```

## Question 1 - Part a

Generating the first five terms of the Legendre polynomial

```python
# call and display the first 5 legendre polynomials starting from P_0
for i in range(5):
    leg = Legendre.basis(i)
    polynomial = leg.convert(kind=Polynomial)
    print(f'P_{i}(x) =', polynomial)
```

## Question 1 - Part b
Calculate the first three Legendre polynomials with Rodrigues' formula

```python
# define a function for Rodrigues' formula
def rodrigues(n: int):
    x = sp.Symbol('x')
    derive = (x**2 - 1)**n
    formula = 1 / (2**n * math.factorial(n)) * sp.diff(derive, x, n)
    print(f'P_{i}(x) =', formula)
```

```python
for i in range(3):
    rodrigues(i)
```

## Question 2
Legendre Polynomial Series for the Sine Function

```python
def legendre(deg: int, z_vals: int, func: np.ndarray):
    leg = Legendre.fit(z_vals, func, deg) # approximation of the sin function
    polynomial = leg.convert(kind=Polynomial)
    print(f'P_{i}(x) =', polynomial)
    func_approx = leg(z_vals)
    return z_vals, func, func_approx
```

```python
z_vals = np.linspace(-1, 1, 100) # fit from -1 < z < 1
func_one = np.sin(np.pi * z_vals)

for i in range(6):
    plt.cla()
    z_vals, func_one, func_approx = legendre(i, z_vals, func_one)
    plt.plot(z_vals, func_one, color='Red', label='Original Function')
    plt.plot(z_vals, func_approx, color='Blue', label='Function Approximation')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
z_vals = np.linspace(-2, 2, 100) # fit from -1 < z < 1
func_one = np.sin(np.pi * z_vals)

for i in range(10):
    plt.cla()
    z_vals, func_one, func_approx = legendre(i, z_vals, func_one)
    plt.plot(z_vals, func_one, color='Red', label='Original Function')
    plt.plot(z_vals, func_approx, color='Blue', label='Function Approximation')
    plt.grid(True)
    plt.ylim(-2, 2)
    plt.legend()
    plt.show()
```

For this approximation, I found including 6 Legendre terms made a good approximation for f(z) from -1 to 1. I found including 10 Legendre terms made a good approximation for f(z) from -2 to 2. What I feel makes a good approximation is that the approximation line is fully covering the line of the original function from the specified bounds.
